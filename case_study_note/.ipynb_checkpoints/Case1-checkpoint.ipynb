## Example 1 – Integrating CRISPR screening datasets

Previously, we performed loss-of-function CRISPR screening analysis to figure out the regulator of CXCL10 in three cell lines (BxPC3, MiaPaca2 and U937). 

CXCL10 is one of the chemokines responsible for T-cell migration to the tumour microenvironment, so we explored which genes are involved in regulating CXCL10 and have the potential to control T-cell migration. 

The CRISPR screening data is analysed by [MaGeCK](https://genomebiology.biomedcentral.com/articles/10.1186/s13059-014-0554-4).

### 1. Integrate two pancriatic cell line datasets

BxPC3 and MiaPaca2 are both pancreatic cell lines, so let's start from detecting the overlap of these two datasets in the “Overlap with two data” sections. 

```python
from IPython.display import Video
Video("video/video1.mp4", width=800)
```

We got 14 genes knockout of which significantly led to downregulation of CXCL10 in two pancreatic cell lines. 

### 2. Integrate the data of myeloid cell line

Next, to explore if these genes are specific to pancreatic cells or if the same effect can be seen in myeloid cells, we integrate all three datasets in the “Compare across datasets” section. 

By inputting the 14 genes obtained from the first step, let's check the LFC of all the screening data.

```python
Video("video/video2.mp4", width=800)
```

Out of 14 genes, 8 genes (CCNC, FAM60A, KAT2A, MED12, PTPN2, SOCS1, STUB1, ZNF148) showed positive LFC in all the screening datasets. This means they are all potential negative regulators regardless of whether the cell is a tumour cell line or a myeloid cell line. Other 6 genes (EED, HMGB1, MYC, STK11, SUZ12, ZNF217) were inconsistent with pancreatic cell lines results. Knockout of them led to negative LFC, which indicates they are potential positive regulator of CXCL10. 

For the further analysis, let's focus on the 8 genes.

### 2-2 Check the expression in the human pancreatic patient scRNAseq

To see if these 8 genes are really expressed in pancreatic cancer patients and in which cell types, we look into their expressions in scRNAseq obtained from [this paper](https://www.nature.com/articles/s41422-019-0195-y).

OmicsBridge can take scRNAseq data in the "scRNA" section. The data has to be processed by R Seurat package in advance. 

In this section, we will see:
1. An expression distribution in the UMAP plot.
2. A dot plot comparing the gene expressions across specified groups.
3. Pie charts showing the fraction of cells expressing or non-expressing the gene.

```python
Video("video/video2-2.mp4", width=800)
```

We could see that all the 8 genes were well expressed in all cell types in this pancriatics cancer patients.

### 3. Invesitgation with clinical data

To investigate the clinical relevance of these 8 genes, let's utilise TCGA PAAD (Pancreatic adenocarcinoma) in the “Clinical data” section, and explore the following three points:
1. Survival analysis. Overall survival difference between the High and Low expression of each gene.
2. The correlation with the CXCL10 expression. Are they negatively correlated with CXCL10 in the patient bulk RNA?
3. The correlation with the abundance level of T cells. Are they negatively correlated with Tcell migration level?

```python
Video("video/video3.mp4", width=800)
```

We observed:
1. Survival analysis: **FAM60A** was significantly negatively associated with the overall survival in the PAAD patients.
2. Correlation with CXCL10: **STUB1** and **KAT2A** were significantly negatively correlated with the CXCL10 expression in the Bulk RNAseq from the PAAD patients
3. Correlation with the abundance level of T cells: **STUB1** and **KAT2A** (and **FAM60A**, though the effect is very slight) were significantly negatively correlated with the abundance level of cytotoxic lymphocytes estimated by deconvolution.

There are some previous reports for these genes.
- It is already reported that FAM60A affects the malignant progression of some cancers. There is a paper claiming that FAM60A depletion impaired the activities of Akt and β-catenin ([Environmental Toxicology (2022)](https://onlinelibrary.wiley.com/doi/full/10.1002/tox.23496)). But No report for immunology.
- Suppressing STUB1 is reported to enhance IFNg activity and potential for better ICB therapy ([Scientific Reports (2022)](https://www.nature.com/articles/s41598-022-18404-4)).
- Targeting KAT2A can inhibit inflammatory macrophage activation in rheumatoid arthritis ([MedComm (2023)](https://onlinelibrary.wiley.com/doi/10.1002/mco2.306))

These genes are worth further investigating for their potential mechanism for T cell recruitment in pancreatic tumours.
