## Example 2 – Exploring omics datasets from different conditions

We conducted an experiment to investigate the effect of MED12 on the transcriptome by performing RNA sequencing on four cell lines (MDAMB-231, MCF7, BxPC3, and Panc10.05) with MED12 knockout, both with and without cytokine treatment. Each sample (4 conditions x 4 cell lines = 16 samples) has three replicates, resulting in a total of 48 samples.

Here, we present how OmicsBridge can be employed to explore the datasets.

The uploaded data includes the following:
- The count table (CPM normalised) of all genes for all 48 samples.
- The results of the differential gene expression analysis by DESeq2 for each of the four cell lines for the following four comparisons:
  - WT vs MED12 KO (under no treatment)
  - WT vs MED12 KO (under cytokine treatment)
  - Untreated vs Cytokine treated (with WT)
  - Untreated vs Cytokine treated (with MED12 KO)

### 1. PCA plots and 

OmicsBridge enables the creation of swarm plots for comparing gene expression among samples and PCA plots, utilizing either all samples or a selected subset. 

```python
from IPython.display import Video
Video("video/case2_video1.mp4", width=800)
```

We can quickly verify that MED12 expression has significantly decreased in the MED12 KO samples, while several cytokine-responsive genes, including STAT1, were up-regulated in the treated samples. Additionally, PCA plots can be generated effortlessly to highlight unique characteristics across various cell lines or to showcase conditional differences within the same cell line.

### 2. Biological exploration of each data

OmicsBridge makes individual transcriptome analysis much easier. It can create volcano plots highlighting significantly up- and down-regulated genes, perform GO/KEGG analysis or GSEA analysis, and infer the activity of transcription factors. We start by exploring the differentially expressed genes data comparing wild type and MED12 KO in untreated MDAMB-231 cells.

```python
from IPython.display import Video
Video("video/case2_video2.mp4", width=800)
```

GO/KEGG analysis of the top 10% of up- or down-regulated genes, along with GSEA analysis, demonstrated that MED12 knockout strengthened cytokine interactions and PI3K-AKT signalling but reduced pathways related to the cell cycle and MAPK signalling. Consistently, the activity of MYC and E2F family transcription factors diminished, whereas interferon-related transcription factors like STAT1, STAT2, and IRF1 exhibited heightened activity levels.

### 3. Combinding the data from different conditions

Having identified the genes up-regulated by MED12 KO in MDAMB231, we can now save them as a custom gene set for future reference. Next, let us revisit the count data to examine the influence of these genes under different conditions. OmicsBridge enables users to create a heatmap that illustrates the standardised expression levels of the specified genes across samples, grouping them based on their expression patterns into a user-specified number of clusters.

```python
from IPython.display import Video
Video("video/case2_video3.mp4", width=800)
```

We identified genes with a log Fold change of at least 1 and a p-value of 0.05 or lower, resulting in the selection of 569 up-regulated genes from the MED12 KO in MDAMB-231. <br>
Analysing gene expression patterns demonstrated that following cytokine treatment, the upregulation effect of 148 of these genes further increased, whereas that of 236 genes diminished. The KEGG analysis indicates that these are genes related to cytokine interactions and the PI3-Akt signalling pathway, respectively.

### 4. Combinding the data from different cell lines

Next, we analysed the differences in data between various cell lines. First, we started from comparing MDAMB231 and MCF7 because both of them are breast cancer cell lines. <br>
The “Integrate two datasets” section assists users to compare the effect sizes from differnt datasets. Also, OmicsBridge has a tool to generate a simple Venn diagram, which makes users it easy to see the overlapping between the gene lists obtained from two datasets.

```python
from IPython.display import Video
Video("video/case2_video4.mp4", width=800)
```

The integrated plot featuring the LFC from MDAMB231 on the X axis and the LFC from MCF7 on the Y axis did not reveal a clear positive correlation. This indicates that the influence of MED12 knockout has numerous cell line-specific effects. <br>
We identified the up-regulated genes from both cell lines (LFC >= 1 & p-value <= 0.05) and measured the overlap. A total of 29 genes were found to be commonly up-regulated following MED12 knockout.

The "Compare across datasets" section enables users to detect the genes consistently ranked in the top or the bottom N% of genes across the selected datasets. Therefore, we investigated the commonly up-regulated genes (ranked top 10%) across all four cell lines to revela the MED12's effect independet of the cell lines. 

```python
from IPython.display import Video
Video("video/case2_video5.mp4", width=800)
```

We detected 19 genes taht were consistently ranked in the top 10% of up-regulated expression across every cell line. <br>
GO (BP) analysis showed that these genes were related to type I interferon, demonstrating MED12's universal function in regulating cytokine responses. regardless of cell lines. 

With OmicsBridge, all these in-depth explorations of the dataset can be seamlessly completed within a single, intuitive interface, making complex analyses more accessible and efficient.
